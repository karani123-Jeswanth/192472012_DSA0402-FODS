```python
import numpy as np

# Item prices and quantities
prices = np.array([2.50, 3.00, 1.20, 5.00])
quantities = np.array([3, 2, 5, 1])

discount_rate = 0.10  # 10% discount
tax_rate = 0.05       # 5% tax

# Calculate subtotal
subtotal = np.sum(prices * quantities)

# Apply discount
discount_amount = subtotal * discount_rate
discounted_total = subtotal - discount_amount

# Apply tax
tax_amount = discounted_total * tax_rate
final_total = discounted_total + tax_amount

print(f"Subtotal: ${subtotal:.2f}")
print(f"Discount (10%): -${discount_amount:.2f}")
print(f"Tax (5%): +${tax_amount:.2f}")
print(f"Total Cost of Purchase: ${final_total:.2f}")
```

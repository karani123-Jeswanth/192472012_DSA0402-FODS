```python
from collections import Counter
import string
import pandas as pd
import matplotlib.pyplot as plt

# Create sample CSV file with proper quoting around text fields
with open('data.csv', 'w') as f:
    f.write("feedback\n")
    f.write('"The product quality is amazing and the delivery was fast."\n')
    f.write('"Terrible service, the item arrived damaged."\n')
    f.write('"Amazing quality and fast delivery. Highly satisfied!"')

df = pd.read_csv('data.csv')

stop_words = {'the', 'and', 'is', 'a', 'an', 'to', 'in', 'it', 'was', 'of', 'for'}

all_text = " ".join(df['feedback'].astype(str)).lower()
cleaned_text = all_text.translate(str.maketrans('', '', string.punctuation))
words = [word for word in cleaned_text.split() if word not in stop_words]

word_freq = Counter(words)

N = 5  # Simulated user input for top N words
top_n = word_freq.most_common(N)

print(f"Top {N} Most Frequent Words:")
for word, freq in top_n:
    print(f"{word}: {freq}")

# Plot bar graph
words_list, freqs_list = zip(*top_n)
plt.figure(figsize=(7, 4))
plt.bar(words_list, freqs_list, color='teal')
plt.title(f'Top {N} Frequent Words in Customer Feedback')
plt.xlabel('Words')
plt.ylabel('Frequency')
plt.show()
```

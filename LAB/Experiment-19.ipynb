```python
import numpy as np
import scipy.stats as stats

# Sample blood pressure reduction data (n = 50 split between drug and placebo)
np.random.seed(42)
drug_group = np.random.normal(loc=12.5, scale=3.0, size=25)
placebo_group = np.random.normal(loc=4.0, scale=2.5, size=25)

def get_ci(data, confidence=0.95):
    n = len(data)
    mean = np.mean(data)
    se = stats.sem(data)
    h = se * stats.t.ppf((1 + confidence) / 2., n - 1)
    return mean, mean - h, mean + h

drug_mean, drug_lower, drug_upper = get_ci(drug_group)
plac_mean, plac_lower, plac_upper = get_ci(placebo_group)

print(f"New Drug Group - Mean: {drug_mean:.2f}, 95% CI: ({drug_lower:.2f}, {drug_upper:.2f})")
print(f"Placebo Group - Mean: {plac_mean:.2f}, 95% CI: ({plac_lower:.2f}, {plac_upper:.2f})")
```

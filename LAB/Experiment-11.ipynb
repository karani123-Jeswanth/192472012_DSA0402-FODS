```python
import matplotlib.pyplot as plt

days = [1, 5, 10, 15, 20, 25, 30]
sales = [200, 450, 300, 600, 550, 800, 750]
months = ['W1', 'W2', 'W3', 'W4']
monthly_sales = [2000, 2500, 2200, 3000]

plt.figure()
plt.plot(days, sales, marker='o')
plt.show()

plt.figure()
plt.scatter(days, sales, color='red')
plt.show()

plt.figure()
plt.bar(months, monthly_sales, color='green')
plt.show()
```

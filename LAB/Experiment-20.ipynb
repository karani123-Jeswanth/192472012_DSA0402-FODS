```python
import numpy as np
from scipy import stats

# Conversion rate data samples for website design A and B
design_a = np.array([0.045, 0.050, 0.042, 0.048, 0.051, 0.046, 0.049])
design_b = np.array([0.058, 0.062, 0.055, 0.060, 0.063, 0.059, 0.061])

# Perform independent two-sample t-test
t_stat, p_value = stats.ttest_ind(design_a, design_b)

print(f"T-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_value:.6f}")

if p_value < 0.05:
    print("Conclusion: There is a statistically significant difference in mean conversion rates between Design A and Design B.")
else:
    print("Conclusion: No statistically significant difference found between Design A and Design B.")
```

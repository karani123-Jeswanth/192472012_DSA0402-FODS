```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
temp = [30, 32, 45, 60, 70, 78]
rainfall = [2.1, 1.8, 3.2, 4.0, 2.5, 1.2]

plt.figure()
plt.plot(months, temp, color='orange', marker='s')
plt.show()

plt.figure()
plt.scatter(months, rainfall, color='blue')
plt.show()
```

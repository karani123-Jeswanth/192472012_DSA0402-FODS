```python
from collections import Counter
import string

reviews = [
    "Great product! Fast shipping and excellent quality.",
    "Good quality, but shipping was a bit slow.",
    "Excellent product, highly recommended!"
]

combined_text = " ".join(reviews).lower()
cleaned_text = combined_text.translate(str.maketrans('', '', string.punctuation))
words = cleaned_text.split()
word_freq = Counter(words)

print(word_freq)
```

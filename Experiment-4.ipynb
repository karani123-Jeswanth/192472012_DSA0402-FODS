```python
import numpy as np

# Sales data for 4 quarters: [Q1, Q2, Q3, Q4]
sales_data = np.array([120000, 145000, 160000, 190000])

# Calculate total sales for the year
total_sales = np.sum(sales_data)

# Calculate percentage increase from Q1 to Q4
q1_sales = sales_data[0]
q4_sales = sales_data[-1]
percentage_increase = ((q4_sales - q1_sales) / q1_sales) * 100

print(f"Total Sales for the Year: ${total_sales:,.2f}")
print(f"Percentage Increase from Q1 to Q4: {percentage_increase:.2f}%")
```

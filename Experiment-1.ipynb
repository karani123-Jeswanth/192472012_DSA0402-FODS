```python
import numpy as np

# Sample 4x4 matrix representing student scores: [Math, Science, English, History]
student_scores = np.array([
    [85, 90, 78, 88],
    [92, 88, 95, 84],
    [76, 80, 82, 79],
    [90, 92, 89, 91]
])

subjects = ['Math', 'Science', 'English', 'History']

# Calculate average score for each subject (column-wise mean)
subject_averages = np.mean(student_scores, axis=0)

# Determine the index of the highest average score
highest_index = np.argmax(subject_averages)
highest_subject = subjects[highest_index]

print("Subject Averages:", dict(zip(subjects, subject_averages)))
print(f"Subject with the highest average score: {highest_subject} ({subject_averages[highest_index]:.2f})")
```

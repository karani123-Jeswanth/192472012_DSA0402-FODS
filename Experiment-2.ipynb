```python
import numpy as np

# Sample 3x3 matrix representing sales data for different products
sales_matrix = np.array([
    [120, 150, 200],
    [90, 110, 130],
    [250, 300, 280]
])

# Calculate the overall average sales/price
average_sales = np.mean(sales_matrix)

print("Sales Matrix:\n", sales_matrix)
print(f"Average sales/price of products: {average_sales:.2f}")
```

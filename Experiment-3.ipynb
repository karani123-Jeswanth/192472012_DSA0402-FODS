```python
import numpy as np

# Columns format: [Bedrooms, Square Footage, Sale Price]
house_data = np.array([
    [3, 1500, 250000],
    [5, 2800, 520000],
    [4, 2000, 350000],
    [6, 3500, 680000],
    [2, 1100, 180000],
    [5, 3000, 550000]
])

# Filter houses with more than 4 bedrooms
filtered_houses = house_data[house_data[:, 0] > 4]

# Calculate average sale price (Column index 2)
if filtered_houses.size > 0:
    avg_sale_price = np.mean(filtered_houses[:, 2])
    print(f"Average sale price for houses with > 4 bedrooms: ${avg_sale_price:,.2f}")
else:
    print("No houses found with more than 4 bedrooms.")
```

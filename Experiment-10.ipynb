```python
import matplotlib.pyplot as plt

# Monthly sales dataset
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [12000, 15000, 13500, 18000, 22000, 21000]

# 1. Line Plot Generation
plt.figure(figsize=(7, 4))
plt.plot(months, sales, marker='o', color='royalblue', linestyle='-', linewidth=2)
plt.title('Monthly Sales Trend (Line Plot)')
plt.xlabel('Month')
plt.ylabel('Sales Amount ($)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# 2. Bar Plot Generation
plt.figure(figsize=(7, 4))
plt.bar(months, sales, color='seagreen', edgecolor='black')
plt.title('Monthly Sales Comparison (Bar Plot)')
plt.xlabel('Month')
plt.ylabel('Sales Amount ($)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

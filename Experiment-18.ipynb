```python
import pandas as pd
import matplotlib.pyplot as plt

# Data from hospital test for 18 adults
data = {
    'age': [23, 23, 27, 27, 39, 41, 47, 49, 50, 52, 54, 54, 56, 57, 58, 58, 60, 61],
    'pct_fat': [9.5, 26.5, 7.8, 17.8, 31.4, 25.9, 27.4, 27.2, 31.2, 34.6, 42.5, 28.8, 33.4, 30.2, 34.1, 32.9, 41.2, 35.7]
}
df = pd.DataFrame(data)

# Calculate statistics
stats = df.agg(['mean', 'median', 'std'])
print(stats)

# Draw boxplots
plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.boxplot(df['age'])
plt.title('Age Boxplot')

plt.subplot(1, 2, 2)
plt.boxplot(df['pct_fat'])
plt.title('%Fat Boxplot')
plt.tight_layout()
plt.show()
```

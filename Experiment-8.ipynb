```python
import pandas as pd

# Sample sales data DataFrame
data = {
    'product_name': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Mouse', 'Keyboard', 'Laptop', 'Headphones', 'Mouse', 'Monitor', 'Webcam'],
    'quantity_sold': [15, 50, 30, 10, 45, 25, 20, 60, 10, 12, 35]
}
sales_df = pd.DataFrame(data)

# Find top 5 products sold the most
top_5_products = (
    sales_df.groupby('product_name')['quantity_sold']
    .sum()
    .reset_index()
    .sort_values(by='quantity_sold', ascending=False)
    .head(5)
)

print("Top 5 Best-Selling Products:")
print(top_5_products.to_string(index=False))
```

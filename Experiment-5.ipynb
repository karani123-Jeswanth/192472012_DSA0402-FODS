```python
import numpy as np

# Fuel efficiency in MPG for different car models
fuel_efficiency = np.array([24.5, 28.0, 31.2, 22.0, 35.5])

# Calculate average fuel efficiency
avg_efficiency = np.mean(fuel_efficiency)

# Determine percentage improvement between Model 1 (index 0) and Model 3 (index 2)
model_a = fuel_efficiency[0]
model_b = fuel_efficiency[2]
percentage_improvement = ((model_b - model_a) / model_a) * 100

print(f"Average Fuel Efficiency: {avg_efficiency:.2f} MPG")
print(f"Percentage Improvement from Model 1 ({model_a} MPG) to Model 3 ({model_b} MPG): {percentage_improvement:.2f}%")
```
